```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
print("Libraries imported successfully for SGD!")
```

```python
# تحميل أو توليد بيانات Fitbit للتدريب
try:
    df = pd.read_csv('heartrate_seconds_merged.csv')
    X_raw = np.linspace(0, 5, len(df))
    Y = df['Value'].values[:len(X_raw)]
except:
    np.random.seed(42)
    X_raw = np.linspace(0, 5, 100)
    Y = 3 * (X_raw ** 3) - 5 * (X_raw ** 2) + 2 * X_raw + np.random.normal(0, 4, 100)

# تجهيز ميزات الانحدار متعدد الحدود (Polynomial Features Degree 3) مع التوحيد
def build_poly_features(X, degree=3):
    X_poly = np.column_stack([X**i for i in range(1, degree + 1)])
    X_poly = (X_poly - np.mean(X_poly, axis=0)) / np.std(X_poly, axis=0)
    return np.c_[np.ones(X_poly.shape[0]), X_poly] # عمود الـ Bias

X_poly = build_poly_features(X_raw, degree=3)
print("Data shape (X_poly):", X_poly.shape)
```

```python
def compute_loss(Y_true, Y_pred):
    m = len(Y_true)
    return (1 / (2 * m)) * np.sum((Y_pred - Y_true) ** 2)

print("Loss function (MSE) defined successfully.")
```

```python
def stochastic_gd_momentum(X, Y, learning_rate=0.01, beta=0.9, epochs=150):
    m, n = X.shape
    weights = np.zeros(n)
    velocity = np.zeros(n)  # متجه الـ Momentum خاصت الـ SGD
    
    loss_history = []
    momentum_history = []
    
    for epoch in range(epochs):
        # خلط البيانات عشوائياً لكل عينة
        indices = np.random.permutation(m)
        X_shuffled = X[indices]
        Y_shuffled = Y[indices]
        
        for i in range(m):
            xi = X_shuffled[i:i+1]
            yi = Y_shuffled[i:i+1]
            
            prediction = np.dot(xi, weights)
            error = prediction - yi
            gradient = np.dot(xi.T, error).flatten()
            
            # تطبيق الـ Momentum للتحديث العشوائي
            velocity = beta * velocity + learning_rate * gradient
            weights -= velocity
            
        # حساب الخسارة لجميع البيانات بعد نهاية الـ Epoch
        epoch_pred = np.dot(X, weights)
        loss_history.append(compute_loss(Y, epoch_pred))
        momentum_history.append(np.mean(np.abs(velocity)))
        
    return weights, loss_history, momentum_history

# تشغيل الخوارزمية
learning_rate = 0.01
epochs = 150
optimal_weights, losses, momentum_vals = stochastic_gd_momentum(X_poly, Y, learning_rate, beta=0.9, epochs=epochs)

print("="*50)
print("--- SGD with Momentum Results ---")
print("Final Weights:", optimal_weights)
print("Final MSE Loss:", losses[-1])
print("="*50)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

# 1. Actual vs Predicted (Polynomial Fit)
axes[0].scatter(X_raw, Y, color='gray', alpha=0.5, label='Actual Data')
axes[0].plot(X_raw, np.dot(X_poly, optimal_weights), color='darkorange', linewidth=2, label='SGD Fit (Degree 3)')
axes[0].set_xlabel('Scaled Time (X)')
axes[0].set_ylabel('Heart Rate (Y)')
axes[0].set_title('Actual vs Predicted - SGD')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# 2. Loss Curve
axes[1].plot(losses, color='darkorange', linewidth=2)
axes[1].set_xlabel('Epochs')
axes[1].set_ylabel('MSE Loss')
axes[1].set_title('Loss Curve over Epochs')
axes[1].grid(True, alpha=0.3)

# 3. Momentum Evolution
axes[2].plot(momentum_vals, color='purple', linewidth=2)
axes[2].set_xlabel('Epochs')
axes[2].set_ylabel('Velocity Magnitude')
axes[2].set_title('SGD Momentum Dynamics')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
